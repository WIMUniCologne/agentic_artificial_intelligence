## 1. Introduction to Multimodality

**Multimodality** refers to the ability to work with data that comes in different forms, such as text, audio, images, and video. Modern LLMs can accept multimodal data as input and generate it as output, enabling more natural and powerful interactions.

### What is Multimodality?

Traditional language models only process text. Multimodal models can:

- **Process images** - Analyze photos, diagrams, screenshots, documents
- **Understand audio** - Transcribe speech, analyze audio content
- **Handle video** - Extract frames, analyze video content
- **Generate multimodal outputs** - Create images, generate audio (depending on model capabilities)

### Why is Multimodality Important?

Multimodality opens up new possibilities for AI applications:

1. **Real-World Applications** - Most real-world data is multimodal (documents with images, videos with audio, etc.)
2. **Better Understanding** - Visual context can help models understand text better
3. **Accessibility** - Can process images of text (OCR), transcribe audio for hearing-impaired users
4. **Content Creation** - Can analyze and generate multimodal content
5. **Domain-Specific Use Cases** - Medical imaging, document analysis, video understanding

### Multimodal vs. Text-Only Models

| Feature | Text-Only Models | Multimodal Models |
|---------|----------------|-------------------|
| Input Types | Text only | Text, images, audio, video |
| Use Cases | Chat, text generation | Image analysis, document understanding, video analysis |
| Complexity | Simpler | More complex, requires specialized models |
| Cost | Generally lower | Can be higher due to processing overhead |
| Availability | Widely available | Limited to specific providers |

### Supported Models

Several providers offer multimodal capabilities:

- **Google Gemini** - Strong image understanding, supports images, audio, video
- **OpenAI GPT-4 Vision** - Excellent image analysis capabilities
- **Anthropic Claude 3** - Good image understanding
- **Other providers** - Various levels of multimodal support

### LangChain/LangGraph Support

LangChain provides standardized interfaces for multimodal data that work across providers:

- **Standard message formats** - Consistent way to represent multimodal content
- **Provider abstraction** - Same code works with different providers
- **Integration with agents** - Multimodal capabilities in LangGraph agents

### 2.3 Practical Example: Analyzing an Image

Let's create a practical example that analyzes an image. We'll use a publicly available image URL:


```python
# Using a sample image URL (replace with your own image URL)
image_url = "https://www.gstatic.com/webp/gallery/1.jpg"

# display the image
from IPython.display import Image
Image(url=image_url)
```

```python
# Example: Analyze an image from a URL
# Create a message with both text and image
message = HumanMessage(
    content=[
        {"type": "text", "text": "Describe this image in detail. What do you see?"},
        {"type": "image_url", "image_url": {"url": image_url}},
    ]
)

# Invoke the model
try:
    response = llm.invoke([message])
    print("Response:")
    print(response.content)
except Exception as e:
    print(f"Error: {e}")
    print("\nNote: Make sure you have a valid image URL and API key configured.")
```

### 2.4 Working with Local Images

To work with local images, we need to encode them as base64:


```python
image_path = ROOT / "exercises/unit_07/tracing_dashboard.png"
Image(url=image_path)
```

```python
def encode_file(path: str) -> str:
    """Encode an file to base64 string."""
    with open(path, "rb") as file:
        return base64.b64encode(file.read()).decode('utf-8')

# Example usage (uncomment and provide a real image path):
base64_image = encode_file(image_path)
 
message = HumanMessage(
    content=[
        {"type": "text", "text": "What's in this image?"},
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/jpeg;base64,{base64_image}"
            }
        },
    ]
 )
 
response = llm.invoke([message])
print(response.content)
```

## 3. Working with Audio

Audio is another important multimodal input. Let's explore how to work with audio in multimodal models.

### 3.1 Audio Input Formats

Audio can be provided in similar ways to images:

- **From URL** - Audio file hosted online
- **From base64** - Encoded audio data
- **From file path** - Local audio files

### 3.2 Audio Processing Use Cases

- **Speech-to-Text** - Transcribe audio to text
- **Audio Analysis** - Understand audio content, sentiment
- **Language Detection** - Identify spoken language
- **Speaker Identification** - Distinguish between speakers
- **Audio Summarization** - Summarize audio content

### 3.3 Example: Audio Transcription

Note: Audio support varies by model. Some models support audio directly, while others may require preprocessing:


```python
# Example: Audio input (format depends on model provider)
# Gemini and some other models support audio input

# base64 encoded audio for /exercises/unit_09/audio_test.mp3
audio_path = ROOT / 'exercises/unit_09/audio_test.mp3'
base64_audio = encode_file(audio_path)

message = HumanMessage(
    content=[
        {"type": "text", "text": "summarize this audio in a sentence"},
        {
            "type": "media",  # Use "media", NOT "file" or "audio"
            "mime_type": "audio/mp3",  # or "audio/wav", "audio/m4a"
            "data": base64_audio,  # base64 encoded audio
        },
    ]
)

# Note: Audio support is model-dependent
# Some models require audio to be converted to text first using a separate service
# Check your model's documentation for specific audio input formats

print("✅ Audio input format example ready!")
```

```python
ai_msg = llm.invoke([message])
ai_msg.pretty_print()
```

## 5. Practical Applications

Let's explore some practical applications of multimodality:

### 5.1 Document Analysis

Multimodal models excel at analyzing documents with images, charts, and text:

```python
# Example: Document analysis
def analyze_document(image_url: str, question: str = None):
    """
    Analyze a document image and extract information.
    """
    if question is None:
        question = "Extract all text and key information from this document. Identify any charts, tables, or diagrams. Display the text in markdown format."
    
    base64_image = encode_file(image_url)

    message = HumanMessage(
        content=[
            {"type": "text", "text": question},
            {"type": "image_url", "image_url": {"url": f"data:image/jpeg;base64,{base64_image}"}},
        ]
    )
    
    response = llm.invoke([message])
    return response.content

# Usage example:
document_url = ROOT / 'exercises/unit_09/document_test.png'
analysis = analyze_document(document_url)
from IPython.display import display, Markdown
display(Markdown(analysis))
```

## 6. Best Practices & Limitations

### 6.1 Best Practices

When working with multimodal models:

1. **Choose the Right Model** - Different models have different multimodal capabilities
2. **Optimize Image Sizes** - Large images increase costs and latency
3. **Use Appropriate Formats** - Use standard image formats (JPEG, PNG)
4. **Provide Context** - Always include text prompts to guide the model
5. **Handle Errors Gracefully** - Multimodal processing can fail for various reasons
6. **Consider Costs** - Multimodal processing is often more expensive than text-only
7. **Respect Privacy** - Be careful with sensitive images (medical, personal data)

### 6.2 Limitations

Multimodal models have several limitations:

- **Cost** - Processing images/audio/video is more expensive than text
- **Latency** - Multimodal processing takes longer than text-only
- **Model Availability** - Not all models support all modalities
- **Quality Varies** - Image understanding quality varies by model
- **Context Limits** - Large images consume more context window
- **Format Support** - Different models support different formats
- **Rate Limits** - API rate limits may be stricter for multimodal requests

### 6.3 When to Use Multimodality

Use multimodal models when:

- ✅ You need to analyze visual content (images, diagrams, charts)
- ✅ You need OCR capabilities (extracting text from images)
- ✅ You're building document understanding systems
- ✅ You need to process user-uploaded content
- ✅ Visual context is important for understanding

Consider text-only models when:

- ❌ You only need text processing
- ❌ Cost is a primary concern
- ❌ Latency is critical
- ❌ You don't have visual content

### 6.4 Model Selection Guide

| Use Case | Recommended Model |
|----------|------------------|
| General image analysis | Gemini, GPT-4 Vision |
| Document OCR | Gemini, GPT-4 Vision |
| Cost-sensitive applications | Gemini Flash |
| High-quality analysis | GPT-4 Vision, Claude 3 |
| Audio processing | Gemini (limited), specialized services |

### 6.5 Tips for Production

1. **Caching** - Cache image analysis results when possible
2. **Preprocessing** - Resize/optimize images before sending
3. **Error Handling** - Implement robust error handling
4. **Monitoring** - Monitor costs and latency
5. **Fallbacks** - Have fallback strategies for unsupported formats
6. **Testing** - Test with various image types and sizes
7. **Documentation** - Document which models support which features
