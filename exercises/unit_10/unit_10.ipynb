## 1. Introduction to Multi-Agent Systems

**Multi-Agent Systems (MAS)** are AI systems composed of multiple autonomous agents that work together to solve complex problems. Each agent has its own capabilities, knowledge, and decision-making processes, and they coordinate through communication and collaboration.

### What is a Multi-Agent System?

A multi-agent system consists of:

- **Multiple Agents** - Each agent is an autonomous entity with its own goals, capabilities, and decision-making logic
- **Coordination Mechanisms** - Agents communicate and coordinate to achieve shared or individual goals
- **Shared or Distributed State** - Agents may share state or maintain isolated state
- **Orchestration** - A supervisor or coordination mechanism manages agent interactions

### Why Use Multi-Agent Systems?

Single agents have limitations:

1. **Limited Expertise** - One agent can't be an expert in everything
2. **Scalability** - Complex tasks may require specialized capabilities
3. **Parallel Processing** - Multiple agents can work on different parts of a problem simultaneously
4. **Modularity** - Easier to maintain and update specialized agents
5. **Robustness** - If one agent fails, others can continue

Multi-agent systems solve these problems by:

- **Specialization** - Each agent focuses on a specific domain or task
- **Collaboration** - Agents work together to solve complex problems
- **Parallelization** - Multiple agents can work simultaneously
- **Modularity** - Easy to add, remove, or update individual agents

### When to Use Multi-Agent Systems

Multi-agent systems are ideal for:

- ✅ **Complex, multi-step tasks** - Tasks that require different expertise at different stages
- ✅ **Specialized domains** - When you need domain-specific knowledge (e.g., research, coding, analysis)
- ✅ **Parallel processing** - When tasks can be broken down and processed in parallel
- ✅ **Modular requirements** - When you need to easily add or remove capabilities
- ✅ **Human-like workflows** - When tasks mirror how humans collaborate (e.g., research team, development team)

### Single Agent vs. Multi-Agent Systems

| Aspect | Single Agent | Multi-Agent System |
|--------|-------------|-------------------|
| **Complexity** | Simpler to build and debug | More complex, requires coordination |
| **Specialization** | General-purpose | Highly specialized agents |
| **Scalability** | Limited by single agent's capabilities | Scales by adding more agents |
| **Parallelization** | Sequential processing | Can process in parallel |
| **Maintenance** | Single point of update | Modular, update individual agents |
| **Use Case** | Simple, focused tasks | Complex, multi-domain tasks |

### Key Concepts in Multi-Agent Systems

1. **Handoffs** - Transferring control from one agent to another
2. **State Sharing** - How agents share information and context
3. **Orchestration** - How agents are coordinated (supervisor, peer-to-peer, etc.)
4. **Subgraphs** - Encapsulating individual agents as subgraphs in LangGraph
5. **Communication Patterns** - How agents communicate (direct, through supervisor, broadcast, etc.)

## 2. Multi-Agent Architectures

There are several common architectures for multi-agent systems. Each has different strengths and use cases.

### 2.1 Supervisor Architecture

In a **supervisor architecture**, a central supervisor agent coordinates specialized worker agents. The supervisor decides which agent should handle each task.

```
┌─────────────┐
│  Supervisor │
│   Agent     │
└──────┬──────┘
       │
       ├──► Worker Agent 1 (Research)
       ├──► Worker Agent 2 (Coding)
       ├──► Worker Agent 3 (Analysis)
       └──► Worker Agent 4 (Writing)
```

**Characteristics:**
- Centralized control and decision-making
- Supervisor routes tasks to appropriate workers
- Workers are specialized and focused
- Clear hierarchy and responsibility

**Use Cases:**
- Task routing and delegation
- Quality control and validation
- Complex workflows with clear stages
- When you need centralized oversight

**Advantages:**
- ✅ Clear control flow
- ✅ Easy to understand and debug
- ✅ Centralized decision-making
- ✅ Good for hierarchical tasks

**Disadvantages:**
- ❌ Supervisor can become a bottleneck
- ❌ Less flexible than peer-to-peer
- ❌ Single point of failure (supervisor)

### 2.2 Network/Swarm Architecture

In a **network/swarm architecture**, agents communicate directly with each other in a peer-to-peer fashion. There's no central coordinator.

```
Agent 1 ◄──► Agent 2 ◄──► Agent 3
   ▲              │              │
   │              ▼              ▼
Agent 4 ◄──► Agent 5 ◄──► Agent 6
```

**Characteristics:**
- Decentralized coordination
- Agents communicate directly
- No single point of control
- More flexible and adaptive

**Use Cases:**
- Collaborative problem-solving
- When agents need to share information
- Distributed decision-making
- Emergent behavior scenarios

**Advantages:**
- ✅ Highly flexible
- ✅ No single point of failure
- ✅ Natural for collaborative tasks
- ✅ Can adapt to changing conditions

**Disadvantages:**
- ❌ More complex to coordinate
- ❌ Harder to debug
- ❌ Potential for conflicts
- ❌ Less predictable behavior

### 2.3 Hierarchical Architecture

A **hierarchical architecture** has multiple levels of agents, with supervisors at different levels managing different groups of workers.

```
        ┌─────────────┐
        │ Top-Level   │
        │ Supervisor  │
        └──────┬──────┘
               │
    ┌──────────┼──────────┐
    │          │          │
┌───▼───┐  ┌──▼───┐  ┌───▼───┐
│ Mid-1 │  │Mid-2 │  │ Mid-3 │
└───┬───┘  └──┬───┘  └───┬───┘
    │         │          │
  Workers   Workers   Workers
```

**Characteristics:**
- Multiple levels of supervision
- Each level manages a subset of agents
- Good for large, complex systems
- Scales well with organization size

**Use Cases:**
- Large organizations or systems
- Multi-level decision-making
- When you need different levels of abstraction
- Enterprise-scale applications

**Advantages:**
- ✅ Scales to large systems
- ✅ Clear organizational structure
- ✅ Different levels of abstraction
- ✅ Good for complex hierarchies

**Disadvantages:**
- ❌ Most complex to implement
- ❌ Can be slow (multiple hops)
- ❌ Harder to debug across levels
- ❌ Overhead of multiple supervisors

### Choosing an Architecture

| Architecture | Best For | Complexity | Scalability |
|-------------|----------|------------|-------------|
| **Supervisor** | Task routing, quality control | Low-Medium | Medium |
| **Network/Swarm** | Collaboration, peer-to-peer | Medium-High | High |
| **Hierarchical** | Large systems, organizations | High | Very High |

In this unit, we'll focus on **Supervisor** and **Network/Swarm** architectures as they're the most commonly used and easiest to understand.

## 3. Agent Communication & Handoffs

In multi-agent systems, agents need to communicate and transfer control. In LangGraph, this is done through a) **handoffs** or b) **tool calls**.

### Multi-agent patterns

| Pattern | How it works | Control flow | Example use case |
|---------|--------------|--------------|------------------|
| **Tool Calling** | A **supervisor** agent calls other agents as **tools**. The "tool" agents do not interact directly with the user; instead, they execute their assigned task and return the results to the supervisor agent. | Centralized. All routing and communication pass through the calling (supervisor) agent. | Task orchestration and structured workflows |
| **Handoffs** | The currently active agent makes a decision to **transfer control** to another agent. When control is transferred, the active agent changes, and the user can then continue interacting directly with the newly active agent. | Decentralized. Agents have the ability to change which agent is currently active and interacting with the user. | Multi-domain conversations and scenarios requiring a specialist takeover |

### Tool calling multi-agent pattern

In **tool calling**, one agent, referred to as the **controller**, treats other agents as *tools* that can be invoked when necessary. The **controller** is responsible for managing orchestration, while the **tool agents** are responsible for performing specific tasks and returning results.

**Flow:**

1. The **controller** receives input and then decides which tool (subagent) to call.
2. The **tool agent** executes its assigned task based on the instructions provided by the controller.
3. The **tool agent** sends its results back to the controller.
4. The **controller** then evaluates the results and decides on the next step in the process or concludes the interaction.

> **Note:** Agents used as tools are generally **not expected** to continue conversation with the user. Their primary role is to perform a task and return results to the controller agent. If there is a need for subagents to directly converse with the user, **handoffs** should be used instead of tool calling.

### What is a Handoff?

A **handoff** is the process of transferring control from one agent to another. This includes:

1. **Transferring the conversation state** - Messages and context
2. **Routing to the correct agent** - Determining which agent should handle the task
3. **Maintaining continuity** - Ensuring the conversation flows naturally

### Handoff Mechanisms in LangGraph

LangGraph provides several ways to implement handoffs:

1. **Direct State Passing** - Pass state directly between agents
2. **Subgraph Invocation** - Call one agent's graph from another
3. **Conditional Routing** - Use conditional edges to route to different agents
4. **Message-Based** - Agents communicate through shared message state

### State Sharing Strategies

There are two main approaches to state management:

1. **Shared State** - All agents share the same state object
   - Pros: Easy to share information, natural conversation flow
   - Cons: All agents see all messages, potential for confusion

2. **Isolated State** - Each agent maintains its own state
   - Pros: Clean separation, agents don't interfere
   - Cons: Need explicit state passing, more complex

In LangGraph, we typically use **shared state** with `MessagesState`, where all agents contribute to the same message history. This creates a natural conversation flow.

### Example: Simple Handoff Pattern

```python
# Pseudo-code for handoff pattern
def supervisor_node(state):
    # Supervisor decides which agent to use
    next_agent = decide_agent(state)
    return {"next_agent": next_agent}

def route_to_agent(state):
    next_agent = state["next_agent"]
    if next_agent == "research":
        return "research_agent"
    elif next_agent == "coding":
        return "coding_agent"
    else:
        return "general_agent"
```

### Choosing a Multi-agent pattern

| Question | Tool Calling | Handoffs |
|----------|--------------|----------|
| Need centralized control over workflow? | ✅ Yes | ❌ No |
| Want agents to interact directly with the user? | ❌ No | ✅ Yes |
| Complex, human-like conversation between specialists? | ❌ Limited | ✅ Strong |

## 4. Building a Supervisor System

Let's build a practical supervisor system with specialized worker agents. We'll create:

1. **Supervisor Agent** - Routes tasks to appropriate workers
2. **Research Agent** - Handles research and information gathering
3. **Coding Agent** - Handles code-related tasks
4. **Analysis Agent** - Handles data analysis and reasoning

### 4.1 Setting Up the Environment

First, let's import the necessary libraries:

````python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from typing import Annotated, Literal
from typing_extensions import TypedDict

# Load environment variables
load_dotenv()

print("✅ Environment setup complete!")
````

### 4.2 Defining the State

We'll use a shared state that all agents can access:


````python
class AgentState(TypedDict):
    """State shared by all agents in the multi-agent system."""
    messages: Annotated[list, add_messages]
    next_agent: str  # Which agent should handle the next step

# Initialize the LLM (we'll use the same LLM for all agents for simplicity)
llm = init_chat_model("gemini-2.5-flash-lite", model_provider="google_genai")

print("✅ State schema and LLM initialized!")
````

### 4.3 Creating Worker Agents

Let's create specialized worker agents. Each agent has a specific role and system prompt:


````python
def research_agent(state: AgentState) -> AgentState:
    """Research agent specializes in finding and summarizing information."""
    messages = state["messages"]
    
    # Add system message for research agent
    system_prompt = """You are a research specialist. Your role is to:
- Find and gather relevant information
- Summarize findings clearly
- Cite sources when possible
- Provide comprehensive research on topics

Be thorough and accurate in your research."""
    
    # Check if system message already exists
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    # Get the last user message
    response = llm.invoke(messages)
    
    return {
        "messages": [response],
        "next_agent": "supervisor"  # Return to supervisor after task
    }

def coding_agent(state: AgentState) -> AgentState:
    """Coding agent specializes in writing and explaining code."""
    messages = state["messages"]
    
    system_prompt = """You are a coding specialist. Your role is to:
- Write clean, well-documented code
- Explain code clearly
- Debug and fix issues
- Follow best practices

Always provide working code examples with explanations."""
    
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    response = llm.invoke(messages)
    
    return {
        "messages": [response],
        "next_agent": "supervisor"
    }

def analysis_agent(state: AgentState) -> AgentState:
    """Analysis agent specializes in data analysis and reasoning."""
    messages = state["messages"]
    
    system_prompt = """You are an analysis specialist. Your role is to:
- Analyze data and information
- Provide logical reasoning
- Break down complex problems
- Draw conclusions from evidence

Be methodical and clear in your analysis."""
    
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    response = llm.invoke(messages)
    
    return {
        "messages": [response],
        "next_agent": "supervisor"
    }

print("✅ Worker agents created!")
````

### 4.4 Creating the Supervisor / Controller / Orchestrator Agent as Router

The supervisor agent decides which worker agent should handle each task:

````python
from pydantic import BaseModel

# use basemodel to define the schema of the supervisor decision
class SupervisorDecision(BaseModel):
    """Schema for supervisor agent decision. Only allows one word: 'research', 'coding', 'analysis', or 'end'."""
    next_agent: Literal["research", "coding", "analysis", "end"]

def supervisor_agent(state: AgentState) -> AgentState:
    """Supervisor agent routes tasks to appropriate worker agents."""
    messages = state["messages"]
    
    # Get the last user message
    last_message = messages[-1].content if messages else ""
    
    system_prompt = """You are a supervisor agent coordinating a team of specialists:
- research_agent: For research, and fact-finding tasks. Not for coding or analysis.
- coding_agent: For programming, code writing, debugging, and technical implementation
- analysis_agent: For data analysis, reasoning, problem-solving, and logical thinking

Based on the user's request, decide which agent should handle it. Respond with ONLY one word: 'research', 'coding', or 'analysis', or 'end' if the task is complete."""
    
    # Create a prompt for the supervisor to decide
    decision_prompt = f"""User request: {last_message}

Which agent should handle this? Respond with ONLY one word: 'research', 'coding', 'analysis', or 'end'."""
    
    decision_messages = [
        SystemMessage(content=system_prompt),
        HumanMessage(content=decision_prompt)
    ]
    structured_decision_llm = llm.with_structured_output(SupervisorDecision)
    decision = structured_decision_llm.invoke(decision_messages)
    next_agent = decision.next_agent.strip().lower()
    print(f"Supervisor decided to route to: {next_agent}")
    
    # Validate and default to supervisor if invalid
    if next_agent not in ["research", "coding", "analysis", "end"]:
        next_agent = "research"  # Default fallback
    
    return {
        "messages": [],  # Supervisor doesn't add messages, just routes
        "next_agent": next_agent
    }

print("✅ Supervisor agent created!")
````

### 4.5 Building the Multi-Agent Graph

Now let's build the graph that connects all agents:

````python
def route_to_agent(state: AgentState) -> Literal["research_agent", "coding_agent", "analysis_agent", "supervisor", "__end__"]:
    """Route to the appropriate agent based on supervisor's decision."""
    next_agent = state.get("next_agent", "supervisor")
    
    if next_agent == "research":
        return "research_agent"
    elif next_agent == "coding":
        return "coding_agent"
    elif next_agent == "analysis":
        return "analysis_agent"
    elif next_agent == "end":
        return "__end__"
    else:
        return "supervisor"  # Default back to supervisor

# Build the graph
workflow = StateGraph(AgentState)

# Add nodes
workflow.add_node("supervisor", supervisor_agent)
workflow.add_node("research_agent", research_agent)
workflow.add_node("coding_agent", coding_agent)
workflow.add_node("analysis_agent", analysis_agent)

# Set entry point
workflow.set_entry_point("supervisor")

# Add conditional routing from supervisor
workflow.add_conditional_edges(
    "supervisor",
    route_to_agent,
    {
        "research_agent": "research_agent",
        "coding_agent": "coding_agent",
        "analysis_agent": "analysis_agent",
        "__end__": END,
        "supervisor": "supervisor"
    }
)

# All worker agents return to supervisor
workflow.add_edge("research_agent", END)
workflow.add_edge("coding_agent", END)
workflow.add_edge("analysis_agent", END)

# Compile the graph
app = workflow.compile()

print("✅ Multi-agent supervisor system built!")
print("\nGraph structure:")
print("START → supervisor → [research_agent | coding_agent | analysis_agent] → supervisor → END")
````

### 4.7 Understanding the Flow

The supervisor system works as follows:

1. **User sends a message** → Enters the graph at `supervisor` node
2. **Supervisor analyzes the request** → Decides which agent should handle it
3. **Routes to worker agent** → Research, Coding, or Analysis agent processes the request
4. **Worker responds** → Adds response to messages and sets `next_agent` to "supervisor"
5. **Returns to supervisor** → Supervisor decides if task is complete or needs another agent
6. **Loop continues** → Until supervisor decides to end

This creates a flexible system where:
- The supervisor can route complex multi-step tasks to different agents
- Each agent specializes in its domain
- The conversation state is maintained across all agents
- The supervisor can coordinate multiple agents for complex tasks

### Key Takeaways

- ✅ **Specialization** - Each agent has a focused role
- ✅ **Centralized Routing** - Supervisor makes routing decisions
- ✅ **Shared State** - All agents share the same message history
- ✅ **Flexible** - Easy to add new worker agents
- ✅ **Maintainable** - Each agent is independent and can be updated separately

### Using the Tool calling pattern

In the **tool calling pattern**, the supervisor agent treats worker agents as tools that can be invoked. This provides centralized control where the supervisor orchestrates the workflow by calling subagents as needed.

**Key differences from handoffs:**
- Subagents are called as tools, not as separate nodes in the graph
- The supervisor maintains control throughout the conversation
- Subagents return results to the supervisor, not directly to the user
- Simpler graph structure with just a supervisor node and a tool execution node

````python
from langchain_core.tools import tool
from langgraph.graph import MessagesState
from langgraph.prebuilt import ToolNode
from langgraph.graph import StateGraph, START, END

# Define state that extends MessagesState (which includes messages)
# AgentState is essentially MessagesState, so we use MessagesState directly
class CustomState(MessagesState):
    """State for the tool-calling supervisor system."""
    pass

# Create subagent graphs (these will be invoked as tools)
def create_research_subgraph():
    """Create a subgraph for the research agent."""
    class ResearchState(MessagesState):
        pass
    
    def research_node(state: ResearchState):
        messages = state["messages"]
        system_prompt = """You are a research specialist. Your role is to:
- Find and gather relevant information
- Summarize findings clearly
- Cite sources when possible
- Provide comprehensive research on topics

Be thorough and accurate in your research."""
        
        if not any(isinstance(msg, SystemMessage) for msg in messages):
            messages = [SystemMessage(content=system_prompt)] + messages
        
        response = llm.invoke(messages)
        return {"messages": [response]}
    
    graph = StateGraph(ResearchState)
    graph.add_node("research", research_node)
    graph.set_entry_point("research")
    graph.add_edge("research", END)
    return graph.compile()

def create_coding_subgraph():
    """Create a subgraph for the coding agent."""
    class CodingState(MessagesState):
        pass
    
    def coding_node(state: CodingState):
        messages = state["messages"]
        system_prompt = """You are a coding specialist. Your role is to:
- Write clean, well-documented code
- Explain code clearly
- Debug and fix issues
- Follow best practices

Always provide working code examples with explanations."""
        
        if not any(isinstance(msg, SystemMessage) for msg in messages):
            messages = [SystemMessage(content=system_prompt)] + messages
        
        response = llm.invoke(messages)
        return {"messages": [response]}
    
    graph = StateGraph(CodingState)
    graph.add_node("coding", coding_node)
    graph.set_entry_point("coding")
    graph.add_edge("coding", END)
    return graph.compile()

def create_analysis_subgraph():
    """Create a subgraph for the analysis agent."""
    class AnalysisState(MessagesState):
        pass
    
    def analysis_node(state: AnalysisState):
        messages = state["messages"]
        system_prompt = """You are an analysis specialist. Your role is to:
- Analyze data and information
- Provide logical reasoning
- Break down complex problems
- Draw conclusions from evidence

Be methodical and clear in your analysis."""
        
        if not any(isinstance(msg, SystemMessage) for msg in messages):
            messages = [SystemMessage(content=system_prompt)] + messages
        
        response = llm.invoke(messages)
        return {"messages": [response]}
    
    graph = StateGraph(AnalysisState)
    graph.add_node("analysis", analysis_node)
    graph.set_entry_point("analysis")
    graph.add_edge("analysis", END)
    return graph.compile()

# Create the subagent graphs
research_subgraph = create_research_subgraph()
coding_subgraph = create_coding_subgraph()
analysis_subgraph = create_analysis_subgraph()

print("✅ Subagent graphs created!")
````

````python
# Define tools that wrap the subagent graphs
# These tools don't need ToolRuntime since they only use the query parameter
@tool(
    "research_agent",
    description="Use this tool for research, fact-finding, and information gathering tasks. Not for coding or analysis."
)
def call_research_agent(query: str) -> str:
    """Call the research agent subgraph to handle research tasks."""
    # Create input for the research subagent
    # Pass the query - the subagent will handle adding system prompts
    subagent_input = {
        "messages": [HumanMessage(content=query)]
    }
    
    # Invoke the research subgraph
    result = research_subgraph.invoke(subagent_input)
    
    # Return the last message content from the subagent
    return result["messages"][-1].content

@tool(
    "coding_agent",
    description="Use this tool for programming, code writing, debugging, and technical implementation tasks."
)
def call_coding_agent(query: str) -> str:
    """Call the coding agent subgraph to handle coding tasks."""
    subagent_input = {
        "messages": [HumanMessage(content=query)]
    }
    
    result = coding_subgraph.invoke(subagent_input)
    return result["messages"][-1].content

@tool(
    "analysis_agent",
    description="Use this tool for data analysis, reasoning, problem-solving, and logical thinking tasks."
)
def call_analysis_agent(query: str) -> str:
    """Call the analysis agent subgraph to handle analysis tasks."""
    subagent_input = {
        "messages": [HumanMessage(content=query)]
    }
    
    result = analysis_subgraph.invoke(subagent_input)
    return result["messages"][-1].content

# Collect all tools
tools = [call_research_agent, call_coding_agent, call_analysis_agent]

print("✅ Tools created for subagents!")
````

````python
# Create supervisor agent that uses the tools
def supervisor_node(state: CustomState) -> CustomState:
    """Supervisor agent that can call subagent tools."""
    messages = state["messages"]
    
    system_prompt = """You are a supervisor agent coordinating a team of specialists:
- research_agent: For research, fact-finding, and information gathering tasks. Not for coding or analysis.
- coding_agent: For programming, code writing, debugging, and technical implementation
- analysis_agent: For data analysis, reasoning, problem-solving, and logical thinking

When a user asks a question, decide which specialist tool to use and call it. 
After getting the result from the tool, synthesize the response for the user.
You can call multiple tools if needed to fully answer the user's question."""
    
    # Add system message if not present
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    # Create LLM with tools bound
    supervisor_llm = llm.bind_tools(tools)
    
    # Get response from supervisor (may include tool calls)
    response = supervisor_llm.invoke(messages)
    
    return {"messages": [response]}

# Build the tool-calling graph
tool_calling_workflow = StateGraph(CustomState)

# Add nodes
tool_calling_workflow.add_node("supervisor", supervisor_node)
tool_calling_workflow.add_node("tools", ToolNode(tools))

# Set entry point
tool_calling_workflow.set_entry_point("supervisor")

# Add conditional routing: if supervisor calls tools, go to tools node, otherwise end
def should_continue(state: CustomState) -> Literal["tools", "end"]:
    """Check if the supervisor called any tools."""
    messages = state["messages"]
    last_message = messages[-1]
    
    # Check if the last message has tool calls
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        return "tools"
    return "end"

tool_calling_workflow.add_conditional_edges(
    "supervisor",
    should_continue,
    {
        "tools": "tools",
        "end": END
    }
)

# After tools execute, return to supervisor
tool_calling_workflow.add_edge("tools", "supervisor")

# Compile the graph
tool_calling_app = tool_calling_workflow.compile()

print("✅ Tool-calling multi-agent system built!")
print("\nGraph structure:")
print("START → supervisor → [tools (if needed)] → supervisor → END")
````

### Key Differences: Handoff vs Tool Calling

**Handoff Pattern (Previous Implementation):**
- Each agent is a separate node in the graph
- Agents transfer control to each other via conditional routing
- More complex graph structure with multiple nodes and edges
- Agents can directly interact with the user
- State routing is explicit through `next_agent` field

**Tool Calling Pattern (Current Implementation):**
- Subagents are encapsulated as tools, not graph nodes
- Supervisor maintains control throughout the conversation
- Simpler graph: just supervisor → tools → supervisor
- Subagents return results to supervisor, not directly to user
- Supervisor synthesizes and presents final response
- More centralized control and easier to manage

**When to Use Each:**
- **Handoffs**: When you need agents to directly interact with users, multi-domain conversations, or when agents need to take over the conversation
- **Tool Calling**: When you need centralized orchestration, task-based workflows, or when the supervisor should always be in control

## 5. Using Subgraphs for Agent Encapsulation

In LangGraph, we can encapsulate individual agents as **subgraphs**. This provides better modularity and allows agents to be reused across different multi-agent systems.

### What are Subgraphs?

Subgraphs are self-contained graphs that can be used as nodes in a larger graph. They:
- Encapsulate agent logic
- Can have their own state schema
- Can be reused in multiple parent graphs
- Provide better organization and debugging

### 5.1 Creating Agent Subgraphs

Let's refactor our worker agents to use subgraphs. First, we'll create individual agent graphs:

````python
from langgraph.graph import MessagesState

# Define a simple state for individual agents (just messages)
class AgentSubgraphState(MessagesState):
    pass

def create_research_subgraph():
    """Create a subgraph for the research agent."""
    def research_node(state: AgentSubgraphState):
        messages = state["messages"]
        system_prompt = """You are a research specialist. Your role is to:
- Find and gather relevant information
- Summarize findings clearly
- Cite sources when possible
- Provide comprehensive research on topics

Be thorough and accurate in your research."""
        
        if not any(isinstance(msg, SystemMessage) for msg in messages):
            messages = [SystemMessage(content=system_prompt)] + messages
        
        response = llm.invoke(messages)
        return {"messages": [response]}
    
    graph = StateGraph(AgentSubgraphState)
    graph.add_node("research", research_node)
    graph.set_entry_point("research")
    graph.add_edge("research", END)
    
    return graph.compile()

def create_coding_subgraph():
    """Create a subgraph for the coding agent."""
    def coding_node(state: AgentSubgraphState):
        messages = state["messages"]
        system_prompt = """You are a coding specialist. Your role is to:
- Write clean, well-documented code
- Explain code clearly
- Debug and fix issues
- Follow best practices

Always provide working code examples with explanations."""
        
        if not any(isinstance(msg, SystemMessage) for msg in messages):
            messages = [SystemMessage(content=system_prompt)] + messages
        
        response = llm.invoke(messages)
        return {"messages": [response]}
    
    graph = StateGraph(AgentSubgraphState)
    graph.add_node("coding", coding_node)
    graph.set_entry_point("coding")
    graph.add_edge("coding", END)
    
    return graph.compile()

def create_analysis_subgraph():
    """Create a subgraph for the analysis agent."""
    def analysis_node(state: AgentSubgraphState):
        messages = state["messages"]
        system_prompt = """You are an analysis specialist. Your role is to:
- Analyze data and information
- Provide logical reasoning
- Break down complex problems
- Draw conclusions from evidence

Be methodical and clear in your analysis."""
        
        if not any(isinstance(msg, SystemMessage) for msg in messages):
            messages = [SystemMessage(content=system_prompt)] + messages
        
        response = llm.invoke(messages)
        return {"messages": [response]}
    
    graph = StateGraph(AgentSubgraphState)
    graph.add_node("analysis", analysis_node)
    graph.set_entry_point("analysis")
    graph.add_edge("analysis", END)
    
    return graph.compile()

# Create the subgraphs
research_subgraph = create_research_subgraph()
coding_subgraph = create_coding_subgraph()
analysis_subgraph = create_analysis_subgraph()

print("✅ Agent subgraphs created!")
````

````python
# Wrapper functions to invoke subgraphs
def invoke_research_subgraph(state: AgentState) -> AgentState:
    """Invoke the research agent subgraph."""
    result = research_subgraph.invoke(state)
    return {
        "messages": result["messages"],
        "next_agent": "supervisor"
    }

def invoke_coding_subgraph(state: AgentState) -> AgentState:
    """Invoke the coding agent subgraph."""
    result = coding_subgraph.invoke(state)
    return {
        "messages": result["messages"],
        "next_agent": "supervisor"
    }

def invoke_analysis_subgraph(state: AgentState) -> AgentState:
    """Invoke the analysis agent subgraph."""
    result = analysis_subgraph.invoke(state)
    return {
        "messages": result["messages"],
        "next_agent": "supervisor"
    }

# Build the supervisor graph with subgraphs
supervisor_workflow = StateGraph(AgentState)

# Add nodes
supervisor_workflow.add_node("supervisor", supervisor_agent)
supervisor_workflow.add_node("research_agent", invoke_research_subgraph)
supervisor_workflow.add_node("coding_agent", invoke_coding_subgraph)
supervisor_workflow.add_node("analysis_agent", invoke_analysis_subgraph)

# Set entry point
supervisor_workflow.set_entry_point("supervisor")

# Add conditional routing
supervisor_workflow.add_conditional_edges(
    "supervisor",
    route_to_agent,
    {
        "research_agent": "research_agent",
        "coding_agent": "coding_agent",
        "analysis_agent": "analysis_agent",
        "__end__": END,
        "supervisor": "supervisor"
    }
)

# All worker agents return to supervisor
supervisor_workflow.add_edge("research_agent", "supervisor")
supervisor_workflow.add_edge("coding_agent", "supervisor")
supervisor_workflow.add_edge("analysis_agent", "supervisor")

# Compile
supervisor_app = supervisor_workflow.compile()

print("✅ Supervisor system with subgraphs built!")
````

## 6. Building a Network/Swarm System

Now let's build a network/swarm system where agents communicate directly with each other. This is useful for collaborative problem-solving.

### 6.1 Network Architecture

In a network system, agents can:
- Communicate directly with each other
- Share information peer-to-peer
- Collaborate on tasks
- Make decisions collectively

Let's create a simple network with three agents that can communicate:

````python
class NetworkState(TypedDict):
    """State for network/swarm system."""
    messages: Annotated[list, add_messages]
    current_agent: str  # Which agent is currently active
    round_count: int  # Track conversation rounds

def agent_a(state: NetworkState) -> NetworkState:
    """Agent A: Research and information gathering specialist."""
    messages = state["messages"]
    round_count = state.get("round_count", 0)
    
    system_prompt = """You are Agent A, a research specialist. 
You gather information and share findings with other agents.
When you receive input, research the topic and share your findings.
If another agent has already responded, build upon their work."""
    
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    # Add context about the network
    context = f"This is round {round_count + 1} of the network conversation."
    messages.append(HumanMessage(content=context))
    
    response = llm.invoke(messages)
    
    return {
        "messages": [response],
        "current_agent": "agent_b",  # Pass to next agent
        "round_count": round_count + 1
    }

def agent_b(state: NetworkState) -> NetworkState:
    """Agent B: Analysis and reasoning specialist."""
    messages = state["messages"]
    round_count = state.get("round_count", 0)
    
    system_prompt = """You are Agent B, an analysis specialist.
You analyze information provided by other agents and add insights.
Build upon what previous agents have said and provide analysis."""
    
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    context = f"This is round {round_count + 1}. Previous agents have shared their thoughts."
    messages.append(HumanMessage(content=context))
    
    response = llm.invoke(messages)
    
    return {
        "messages": [response],
        "current_agent": "agent_c",
        "round_count": round_count + 1
    }

def agent_c(state: NetworkState) -> NetworkState:
    """Agent C: Synthesis and summary specialist."""
    messages = state["messages"]
    round_count = state.get("round_count", 0)
    
    system_prompt = """You are Agent C, a synthesis specialist.
You synthesize information from all agents and provide a comprehensive summary.
Combine insights from previous agents and add final thoughts."""
    
    if not any(isinstance(msg, SystemMessage) for msg in messages):
        messages = [SystemMessage(content=system_prompt)] + messages
    
    context = f"This is round {round_count + 1}. Synthesize all previous contributions."
    messages.append(HumanMessage(content=context))
    
    response = llm.invoke(messages)
    
    # After agent C, decide if we should continue or end
    if round_count >= 2:  # After 3 rounds (A, B, C), end
        next_agent = "end"
    else:
        next_agent = "agent_a"  # Start another round
    
    return {
        "messages": [response],
        "current_agent": next_agent,
        "round_count": round_count + 1
    }

print("✅ Network agents created!")
````

### 6.2 Building the Network Graph

Now let's build the network graph where agents pass control to each other:

````python
def route_network(state: NetworkState) -> Literal["agent_a", "agent_b", "agent_c", "__end__"]:
    """Route to the next agent in the network."""
    current_agent = state.get("current_agent", "agent_a")
    round_count = state.get("round_count", 0)
    
    if current_agent == "end" or round_count >= 3:
        return "__end__"
    
    return current_agent

# Build the network graph
network_workflow = StateGraph(NetworkState)

# Add nodes
network_workflow.add_node("agent_a", agent_a)
network_workflow.add_node("agent_b", agent_b)
network_workflow.add_node("agent_c", agent_c)

# Set entry point
network_workflow.set_entry_point("agent_a")

# Add conditional routing
network_workflow.add_conditional_edges(
    "agent_a",
    lambda state: "agent_b",
    {"agent_b": "agent_b"}
)

network_workflow.add_conditional_edges(
    "agent_b",
    lambda state: "agent_c",
    {"agent_c": "agent_c"}
)

network_workflow.add_conditional_edges(
    "agent_c",
    route_network,
    {
        "agent_a": "agent_a",
        "__end__": END
    }
)

# Compile
network_app = network_workflow.compile()

print("✅ Network/swarm system built!")
print("\nNetwork flow: agent_a → agent_b → agent_c → (loop or end)")
````

### 6.4 Network vs Supervisor Comparison

| Aspect | Supervisor System | Network System |
|--------|------------------|----------------|
| **Control** | Centralized (supervisor) | Decentralized (peer-to-peer) |
| **Routing** | Supervisor decides | Agents pass control sequentially |
| **Flexibility** | Less flexible | More flexible |
| **Complexity** | Lower | Higher |
| **Use Case** | Task routing, delegation | Collaboration, brainstorming |
| **Debugging** | Easier (clear hierarchy) | Harder (distributed) |
| **Scalability** | Medium | High (can add more agents easily) |

### When to Use Each

**Use Supervisor System when:**
- You need centralized control and routing
- Tasks have clear categories that map to agents
- You want predictable, structured workflows
- Quality control and oversight are important

**Use Network System when:**
- Agents need to collaborate and build on each other's work
- You want emergent, creative solutions
- Tasks benefit from multiple perspectives
- You need flexible, adaptive workflows

## 7. Best Practices & Common Patterns

### 7.1 State Management Best Practices

1. **Use Shared State for Conversation Flow**
   - Keep messages in shared state for natural conversation
   - Use `MessagesState` or `Annotated[list, add_messages]` for message handling

2. **Isolate Agent-Specific Data**
   - Use separate fields for agent-specific state
   - Don't pollute shared state with agent internals

3. **Clear State Schema**
   - Define TypedDict for type safety
   - Document what each field is used for

### 7.2 Agent Design Best Practices

1. **Clear Responsibilities**
   - Each agent should have a single, well-defined purpose
   - Avoid overlapping responsibilities

2. **Consistent Interfaces**
   - All agents should use the same state schema
   - Use consistent naming conventions

3. **Error Handling**
   - Handle errors gracefully
   - Provide fallback mechanisms

4. **System Prompts**
   - Write clear, specific system prompts
   - Include role, responsibilities, and constraints

### 7.3 Common Patterns

**Pattern 1: Supervisor with Fallback**
```python
def supervisor_with_fallback(state):
    try:
        return route_to_agent(state)
    except:
        return "general_agent"  # Fallback
```

**Pattern 2: Agent Voting**
```python
def voting_supervisor(state):
    # Multiple agents vote on decision
    votes = [agent_a_vote(), agent_b_vote(), agent_c_vote()]
    return majority_vote(votes)
```

**Pattern 3: Sequential Pipeline**
```python
# Agents process in sequence, each building on previous
agent_a → agent_b → agent_c → final_response
```

**Pattern 4: Parallel Processing**
```python
# Multiple agents work in parallel, then combine results
[agent_a, agent_b, agent_c] → combine_results → final_response
```

### 7.4 Debugging Tips

1. **Use LangSmith Tracing**
   - Enable tracing to see agent execution
   - Track state changes between agents

2. **Add Logging**
   - Log agent decisions and routing
   - Log state at key points

3. **Visualize the Graph**
   - Use LangGraph Studio to visualize
   - Check graph structure and edges

4. **Test Individual Agents**
   - Test each agent in isolation
   - Verify state transformations

### 7.5 Performance Optimization

1. **Cache Agent Responses**
   - Cache common queries
   - Reuse agent subgraphs

2. **Parallel Execution**
   - Run independent agents in parallel
   - Use async execution when possible

3. **Limit Conversation Rounds**
   - Set maximum rounds to prevent infinite loops
   - Use round_count or step limits

4. **Optimize State Size**
   - Don't store unnecessary data in state
   - Trim old messages if needed
