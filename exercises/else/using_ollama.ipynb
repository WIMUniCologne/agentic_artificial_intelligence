```python
from langchain_ollama import OllamaLLM

# Instantiate the Ollama model
# The `model` parameter should match the name of the model you pulled
llm = OllamaLLM(model="gemma3:4b")

# Invoke the model with a prompt
response = llm.invoke("Hi what's up?")

print(response)
```

```python
# use init_chat_model to create a chat model
from langchain.chat_models import init_chat_model
llm = init_chat_model("gemma3:4b", model_provider="ollama")

# invoke the model with a prompt
response = llm.invoke("Hi what's up?")

print(response)
```
