**What changes when a model gets more than text?**

#### 1.1 · A new word

**Multimodality:** a model works with more than one form of data.

**A mode:** one form of data: text, picture, sound or video.

An email has some text, and it can have an **attachment**. A message to a model now works the same way.

#### 1.2 · What can such models do?

**Pictures:** read photos, diagrams, screenshots and documents.

**Sound:** write down speech, and understand what it says.

**Video:** say what happens in a video.

| | Text-only models | Multimodal models |
|:--|:--|:--|
| **Input** | text | text, pictures, sound, video |
| **Cost** | generally lower | can be higher |
| **Which models?** | nearly all | not every model can do everything |

## Part 3 of 10 · A picture from your computer: base64

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**How can a file travel inside a message?**

```python
# 3.1 · A picture from our project
image_path = ROOT / "exercises/unit_07/tracing_dashboard.png"
Image(filename=image_path)
```

#### 3.2 · Bytes and base64

A file is a long row of **bytes**. A byte is a small number, from 0 to 255.

**Base64** writes any file using only plain characters: letters, digits and two signs. Then it can travel as text.

**It is not a secret code.** Anybody can turn it back.

```python
# 3.3 · A function that writes a file as text
def encode_file(path: str) -> str:
    with open(path, "rb") as file:
        return base64.b64encode(file.read()).decode('utf-8')
```

```python
# 3.4 · The same message, now with our file
base64_image = encode_file(image_path)

message = HumanMessage(
    content=[
        {"type": "text", "text": "What's in this image?"},
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/jpeg;base64,{base64_image}"
            }
        },
    ]
)
```

```python
# 3.5 · Send it
response = llm.invoke([message])
print(response.text)
```

```python
# 3.6 · The file as text
print(base64_image[:300], "...")
print()
print(f"Characters in this text: {len(base64_image):,}")
print(f"Bytes in the file: {image_path.stat().st_size:,}")
```

## Part 7 of 10 · Reading a document

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**What can a model do with a document?**

```python
# 7.1 · A function for documents
def analyze_document(image_url: str, question: str = None):
    if question is None:
        question = ("Extract all text and key information from this document. "
                    "Identify any charts, tables, or diagrams. "
                    "Display the text in markdown format.")

    base64_image = encode_file(image_url)

    message = HumanMessage(
        content=[
            {"type": "text", "text": question},
            {"type": "image_url",
             "image_url": {"url": f"data:image/jpeg;base64,{base64_image}"}},
        ]
    )

    response = llm.invoke([message])
    return response.text
```

```python
# 7.2 · Our document
from pathlib import Path

document_url = Path("unit_09_media/sample_invoice.png")
Image(url=f"data:image/png;base64,{encode_file(document_url)}", width=700)
```

```python
# 7.3 · Read the document
analysis = analyze_document(document_url)

from IPython.display import display, Markdown
display(Markdown(analysis))
```
