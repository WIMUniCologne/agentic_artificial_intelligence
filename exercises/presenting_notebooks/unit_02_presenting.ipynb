## Part 2 of 10 · How a model sees text

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**What does a model see when we type a sentence?**

```python
# 2.1 · The tool and our sentence
from transformers import AutoTokenizer

text_to_tokenize = ("Tokenization is fascinating and the course "
                    "Agentic Artificial Intelligence is exciting!")
```

```python
# 2.2 · A small helper function
def tokenize_text(text_to_tokenize, tokenizer_name = "bert-base-uncased"):
    tokenizer = AutoTokenizer.from_pretrained(tokenizer_name)

    encoded_ids = tokenizer.encode(text_to_tokenize)
    print(f"Original Text: '{text_to_tokenize}'")
    print(f"Token IDs: {encoded_ids}")

    tokens = tokenizer.convert_ids_to_tokens(encoded_ids)
    print(f"Tokens: {tokens}")

    decoded_text = tokenizer.decode(encoded_ids)
    print(f"Decoded Text: '{decoded_text}'")
```

```python
# 2.3 · First tokenizer: BERT
tokenize_text(text_to_tokenize)
```

```python
# 2.4 · Second tokenizer: GPT-2
tokenize_text(text_to_tokenize, 'gpt2')
```

#### 2.5 · What did we just see?

<details>
<summary>Click to show</summary>

A **token** is a piece of text: a word, a part of a word, or a sign.

A **tokenizer** cuts a text into tokens and gives each token a number.

**The model only sees these numbers.** Each model family has its own list: 30,522 tokens for BERT, 50,257 for GPT-2.

</details>

**How does a model write a whole text?**

#### 3.1 · The loop

<img src="../unit_02/AutoregressionSchema.gif" alt="AutoregressionSchema" width="800">

The new token goes back into the input. Then the next round starts.

**EOS** means end of sequence. It is a special token.

#### 3.2 · One round, from the inside

<img src="../unit_02/DecodingFinal.gif" alt="DecodingFinal" width="800">

**Logit:** a raw score for a possible next token.

**Softmax:** turns the scores into probabilities that add up to 1.

**Probability:** 0.35 means 35 out of 100.

The picture shows only five tokens. A real model gives a score to every token in its list.

#### 3.3 · Does a model always take the highest?

<details>
<summary>Click to show</summary>

Not always. Many models **sample**: the probabilities work like a dice with weights.

A likely token comes up often, a less likely one sometimes.

So the same question can give different answers.

</details>

#### 3.4 · One sentence for today

<details>
<summary>Click to show</summary>

**A language model only predicts the next token, again and again. Everything else we build around it.**

</details>

## Part 6 of 10 · Messages and templates

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**How do we send a whole conversation, and how do we reuse a good prompt?**

#### 6.1 · Two kinds of template

| | What it does | Who writes it here |
|:--|:--|:--|
| **Chat template** | turns a list of messages into the one long text that a model reads, with special tokens | the provider, on its side |
| **Prompt template** | a text with gaps, which we fill with our values | we do |

```python
# 6.2 · One more import
from langchain_core.prompts import ChatPromptTemplate
```

```python
# 6.3 · Four messages
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

system_message = SystemMessage(content="You are a helpful assistant that can answer "
                                       "questions and help with tasks.")
human_message_1 = HumanMessage(content="What is the capital of France?")
ai_message = AIMessage(content="The capital of France is Paris.")
human_message_2 = HumanMessage(content="What is the capital of Germany?")
```

```python
# 6.4 · Put them into one prompt
prompt = ChatPromptTemplate.from_messages([system_message, human_message_1,
                                           ai_message, human_message_2])

prompt
```

```python
# 6.5 · Send the conversation
answer = llm.invoke(prompt.invoke({}))
print(answer.text)
```

```python
# 6.6 · A prompt with gaps
from langchain_core.prompts import PromptTemplate

template = """You are an expert {role}. 

Question: {question}

Please provide a detailed answer that includes:
1. A clear explanation
2. Real-world examples
3. Practical implications

Answer:"""

prompt = PromptTemplate(
    input_variables=["role", "question"],
    template=template
)
```

```python
# 6.7 · Fill the gaps
formatted_prompt = prompt.format(
    role="machine learning engineer",
    question="How do transformer models work?"
)

print("📝 Generated Prompt:")
print("=" * 40)
print(formatted_prompt)
print("\n" + "=" * 40)
```

```python
# 6.8 · Send the filled prompt
response = llm.invoke(formatted_prompt)
print("\n🤖 Response:")
print(response.text)
```

## Part 7 of 10 · From text to data

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**How does our program get one number out of an answer?**

#### 7.1 · Text is hard to use

We want a table of films, with title, year and rating.

An answer in plain text can look different every time. So we describe a **form**, and the answer must fit it.

**Pydantic** is a Python library that checks such forms.

```python
# 7.2 · Describe a form
from pydantic import BaseModel

class Person(BaseModel):
    first_name: str
    last_name: str
    age: int
    email: str
```

```python
# 7.3 · Fill the form
person = Person(
    first_name="Alice",
    last_name="Smith",
    age=30,
    email="alice@example.com"
)

print("✅ Person object created successfully!")
print(f"Full name: {person.first_name} {person.last_name}")
print(f"Age: {person.age}")
print(f"Email: {person.email}")
```

```python
# 7.4 · Ready to save
print("\n📋 As dictionary:")
print(person.model_dump())

print("\n📝 As JSON:")
print(person.model_dump_json(indent=2))
```

```python
# 7.5 · An age in quotes
print("Example 1: Type coercion")
person2 = Person(
    first_name="Bob",
    last_name="Jones",
    age="25",
    email="bob@example.com"
)
print(f"✅ Age '25' (string) was converted to {person2.age} (int)")
print(f"Type of age: {type(person2.age)}\n")
```

```python
# 7.6 · An age that is not a number
print("Example 2: Invalid data")
try:
    invalid_person = Person(
        first_name="Charlie",
        last_name="Brown",
        age="not a number",
        email="charlie@example.com"
    )
except Exception as e:
    print(f"❌ Validation error: {e}")
```

```python
# 7.7 · Two more imports
from typing import List
from pydantic import Field
```

```python
# 7.8 · A form for a film review
class MovieReview(BaseModel):
    """A movie review with structured information."""
    title: str = Field(description="The movie title")
    year: int = Field(description="The year the movie was released")
    genre: List[str] = Field(description="List of genres (e.g., Action, Comedy, Drama)")
    rating: float = Field(description="Rating from 0.0 to 10.0")
    summary: str = Field(description="Brief one-sentence summary")
    recommendation: str = Field(description="Would you recommend it? (Yes/No/Maybe)")
```

```python
# 7.9 · Give the form to the model
structured_llm = llm.with_structured_output(MovieReview)

prompt = """Analyze the movie 'The Matrix' and provide a review with the following information:
- Title
- Release year
- Genres
- Your rating (0-10)
- Brief one-sentence summary
- Whether you'd recommend it"""
```

```python
# 7.10 · Ask for a review
print("🎬 Requesting structured movie review from LLM...")
print("=" * 50)

review = structured_llm.invoke(prompt)

print(f"✅ Received structured output!\n")
print(f"Title: {review.title}")
print(f"Year: {review.year}")
print(f"Genre: {', '.join(review.genre)}")
print(f"Rating: {review.rating}/10.0")
print(f"Summary: {review.summary}")
print(f"Recommendation: {review.recommendation}")
```

```python
# 7.11 · What did we get?
print(f"\n📊 Data type: {type(review)}")
print(f"✅ This is a validated MovieReview object, not just text!")

print("\n💾 Converting to different formats:")
print("\n1. As dictionary:")
print(review.model_dump())

print("\n2. As JSON:")
print(review.model_dump_json(indent=2))
```

#### 7.12 · What is checked, and what is not?

<details>
<summary>Click to show</summary>

**Checked:** every field is there and has the right type.

**Not checked:** whether the content is true. A rating of 11, or a wrong year, also fits this form.

</details>

## Part 8 of 10 · A conversation over several calls

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**What does the model get when we talk to it three times?**

```python
# 8.1 · A chain and a buffer
from langchain_classic.memory import ConversationBufferMemory
from langchain_classic.chains import ConversationChain

memory = ConversationBufferMemory()

conversation = ConversationChain(
    llm=llm,
    memory=memory,
    verbose=True
)
```

```python
# 8.2 · First message
print("🗣️ Starting a conversation with memory:")
print("=" * 50)

response1 = conversation.predict(
    input="Hi! My name is Alex and I'm learning about AI agents.")
print(f"User: Hi! My name is Alex and I'm learning about AI agents.")
print(f"AI: {response1}\n")
```

```python
# 8.3 · Second message
response2 = conversation.predict(input="What are the key components I should focus on?")
print(f"User: What are the key components I should focus on?")
print(f"AI: {response2}\n")
```

```python
# 8.4 · Third message
response3 = conversation.predict(input="Can you remind me what my name is?")
print(f"User: Can you remind me what my name is?")
print(f"AI: {response3}\n")
```

```python
# 8.5 · What is stored?
print("🧠 Memory Contents:")
print("=" * 30)
print(memory.buffer)
```

<details>

## Part 9 of 10 · Streaming

<div style="display:flex;gap:6px;margin:0 0 14px 0"><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#1F3A5F"></span><span style="flex:1;height:10px;border-radius:5px;background:#D5DBE3"></span></div>

**Why do chat apps show their answer piece by piece?**

```python
# 9.1 · A second connection
import time

streaming_llm = init_chat_model("gemini-3.5-flash-lite", model_provider="google_genai",
                                max_tokens=2000, streaming=True)
```

```python
# 9.2 · Ask, and print the pieces
print("🌊 Streaming Response Example:")
print("=" * 40)
print("Question: Explain the concept of attention mechanism in transformers in detail.\n")
print("Response (streaming):")
print("-" * 20)

for chunk in streaming_llm.stream(
        "Explain the concept of attention mechanism in transformers in detail."):
    print(chunk.text, end='', flush=True)
    time.sleep(0.1)

print("\n" + "-" * 20)
print("✅ Streaming complete!")
```

**What should you remember from today?**

#### 10.1 · Five things

**1.** A model sees **tokens**: numbers, not letters.

**2.** It predicts the **next token**, again and again, until the end token. Everything else we build around it.

**3.** With **attention**, some earlier tokens count more than others.

**4.** In our code, the answer is **only text**, until we give it a **form**. The form checks types, not truth.

**5.** **Memory** is old messages in the next prompt. **Streaming** shows the pieces as they come.
