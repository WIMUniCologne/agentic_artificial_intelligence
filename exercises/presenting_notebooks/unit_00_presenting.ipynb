**How do we repeat something, and how do we choose?**

````python
# 8.1 · Our first loop
lst = [0,1,'a',3,4,5,6,7,8,9]
for i in lst:
    print(i)
````

````python
# 8.2 · A loop with range
for i in range(1,5,1):
    print('myname')
````

````python
# 8.3 · Earlier we wrote i = 5. And now?
i
````

````python
# 8.4 · A loop with if
for i in range(10):
    if i%2 == 0:
        print(i)
````

````python
# 8.5 · A loop with if and else
for i in range(10):
    # Check if it is odd
    if i%2 != 0:
        print(i, ' is odd')
    # If not odd (even), then do the following
    else:
        print(i, ' is even')
````

**How do we use the same code again?**

````python
# 9.1 · Our first function
def run_simple_loop():
    for i in range(10):
        print(i)
    return None
````

````python
# 9.2 · Call it
run_simple_loop()
````

````python
# 9.3 · A function that takes a number
def run_simple_loopX(x):
    for i in range(x):
        print(i)
    return None
````

````python
# 9.4 · No number: a red error, on purpose
run_simple_loopX()
````

#### 9.5 · How to read a red error

**The red text in step 9.4 is on purpose.** Nothing is broken, and the notebook goes on.

**Read the last line first.** It says what is wrong. The arrow above it shows the line.

````python
# 9.6 · With a number. What is in b?
b = run_simple_loopX(3)
b
````

````python
# 9.7 · A new last line. What is in a?
def run_simple_loopX(x):
    for i in range(x):
        print(i)
    return range(x)

a = run_simple_loopX(3)
a
````
